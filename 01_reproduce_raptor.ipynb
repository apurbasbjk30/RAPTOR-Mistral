```python
paper_id = list(train_data.keys())[0]

paper = train_data[paper_id]

print(paper["title"])
```

```python
def paper_to_text(paper):

    text = ""

    # Title
    text += paper["title"] + "\n\n"

    # Abstract
    text += paper["abstract"] + "\n\n"

    # Full paper
    for section in paper["full_text"]:

        text += section["section_name"] + "\n"

        for para in section["paragraphs"]:
            text += para + "\n"

        text += "\n"

    return text
```

```python
paper_text = paper_to_text(paper)

print(paper_text[:1500])
```

```python
import time

start = time.time()

RA.add_documents(paper_text)

print(f"Tree built in {time.time()-start:.2f} seconds")
```
